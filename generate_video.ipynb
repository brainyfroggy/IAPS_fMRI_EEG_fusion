```python
import os
from PIL import Image, ImageDraw, ImageFont
import numpy as np
import matplotlib.pyplot as plt

# --- Paths ---
left_out_folder = r'N:\Experimental_Data\yujunchen\projects\IAPS_EEG_RSA\frames_inflated_surf\LH_out'
left_in_folder  = r'N:\Experimental_Data\yujunchen\projects\IAPS_EEG_RSA\frames_inflated_surf\LH_in'
right_out_folder= r'N:\Experimental_Data\yujunchen\projects\IAPS_EEG_RSA\frames_inflated_surf\RH_out'
right_in_folder = r'N:\Experimental_Data\yujunchen\projects\IAPS_EEG_RSA\frames_inflated_surf\RH_in'
folders = [left_out_folder, left_in_folder, right_out_folder, right_in_folder]
labels  = ['LH_out', 'LH_in', 'RH_out', 'RH_in']
outdir  = r'N:\Experimental_Data\yujunchen\projects\IAPS_EEG_RSA\frames\merged_frames_inflated'
os.makedirs(outdir, exist_ok=True)

n_frames = 574  # frames indexed 0..573

# --- Font setup ---
font_path = r"C:\Windows\Fonts\arial.ttf"
font_size = 80
try:
    font = ImageFont.truetype(font_path, font_size)
except:
    font = ImageFont.load_default()

# --- helper: center crop to (target_w, target_h) ---
def center_crop(im: Image.Image, target_w: int, target_h: int) -> Image.Image:
    w, h = im.size
    # assume w >= target_w and h >= target_h (true since we use global min)
    left   = (w - target_w) // 2
    top    = (h - target_h) // 2
    right  = left + target_w
    bottom = top + target_h
    return im.crop((left, top, right, bottom))

# --- Step 1: Find smallest width and height across all folders/frames ---
min_w, min_h = None, None
for folder in folders:
    for i in range(n_frames):
        img_path = os.path.join(folder, f'frame_{i:03d}.png')
        with Image.open(img_path) as im:
            w, h = im.size
            if (min_w is None) or (w < min_w):
                min_w = w
            if (min_h is None) or (h < min_h):
                min_h = h

print(f"Smallest width: {min_w}, height: {min_h}")

# --- Step 2: Center-crop (NOT resize) and annotate each image ---
annotated = {}  # store processed images in memory by folder
for folder, label in zip(folders, labels):
    annotated[folder] = []
    for i in range(n_frames):
        img_path = os.path.join(folder, f'frame_{i:03d}.png')
        with Image.open(img_path) as im:
            # Center-crop to (min_w, min_h) instead of resizing
            im = center_crop(im, min_w, min_h)
            draw = ImageDraw.Draw(im)
            text = label
            # text size
            try:
                bbox = draw.textbbox((0, 0), text, font=font)
                text_w = bbox[2] - bbox[0]
                text_h = bbox[3] - bbox[1]
            except AttributeError:
                text_w, text_h = draw.textsize(text, font=font)
            # Draw label at bottom-center of each tile
            draw.text(((min_w - text_w) // 2, min_h - text_h - 10),
                      text, font=font, fill='white')
            annotated[folder].append(im.copy())

def make_colorbar(
    width=200, bar_height=10, vmin=-0.03, vmax=0.06, cmap_name='jet', text_height=50,
    label_font_path=r"C:\Windows\Fonts\arial.ttf", label_font_size=100):
    import matplotlib.pyplot as plt
    import numpy as np
    from PIL import Image, ImageDraw, ImageFont

    total_height = bar_height + text_height

    # Create colorbar image (no text)
    fig, ax = plt.subplots(figsize=(width / 100, total_height / 100), dpi=100)
    fig.subplots_adjust(left=0, right=1, top=1, bottom=0)
    gradient = np.linspace(vmin, vmax, width)
    gradient = np.tile(gradient, (bar_height, 1))
    ax.imshow(gradient, aspect='auto', cmap=cmap_name, origin='lower',
              extent=[0, width, text_height, text_height + bar_height])
    ax.axis('off')
    fig.canvas.draw()
    colorbar_img = np.array(fig.canvas.renderer.buffer_rgba())
    colorbar_img = colorbar_img[..., :3]  # Remove alpha
    plt.close(fig)

    # Add text via PIL
    pil_img = Image.fromarray(colorbar_img)
    draw = ImageDraw.Draw(pil_img)
    try:
        fnt = ImageFont.truetype(label_font_path, label_font_size)
    except:
        fnt = ImageFont.load_default()

    y_label = text_height - 30
    # Left label
    left_label = f'{vmin:.2f}'
    bbox = draw.textbbox((0, 0), left_label, font=fnt)
    tw = bbox[2] - bbox[0]
    draw.text((0, y_label), left_label, font=fnt, fill='white')
    # Right label
    right_label = f'{vmax:.2f}'
    bbox = draw.textbbox((0, 0), right_label, font=fnt)
    tw = bbox[2] - bbox[0]
    draw.text((width - tw, y_label), right_label, font=fnt, fill='white')

    return pil_img

def draw_text_with_outline(draw, xy, text, font, fill='white', outline='black', width=2):
    x, y = xy
    for dx, dy in [(-width,0),(width,0),(0,-width),(0,width),
                   (-width,-width),(-width,width),(width,-width),(width,width)]:
        draw.text((x+dx, y+dy), text, font=font, fill=outline)
    draw.text((x, y), text, font=font, fill=fill)

margin = 30

# --- Step 4: Merge 4 tiles + colorbar + timestamp ---
for i in range(n_frames):
    img_left  = annotated[left_out_folder][i]
    img_right = annotated[right_out_folder][i]
    img_mid   = annotated[left_in_folder][i]
    img_top   = annotated[right_in_folder][i]
    w, h = img_left.size  # == (min_w, min_h)

    merged = Image.new('RGBA', (w*2, h*2), (0,0,0,0))
    merged.paste(img_left,  (0,   0))
    merged.paste(img_right, (w,   0))
    merged.paste(img_mid,   (0,   h))
    merged.paste(img_top,   (w,   h))

    # Colorbar (centered top)
    colorbar_width = merged.width // 4
    colorbar_bar_height = 10
    colorbar_text_height = 40
    colorbar = make_colorbar(
        width=colorbar_width,
        bar_height=colorbar_bar_height,
        text_height=colorbar_text_height,
        vmin=-0.03,
        vmax=0.05,
        label_font_path=font_path,
        label_font_size=40
    )
    bar_x = (merged.width - colorbar_width) // 2
    bar_y = 10
    merged.paste(colorbar, (bar_x, bar_y))

    # Timestamp (top-right): last frame is 2000ms, step -4ms backwards
    last_idx = n_frames - 1
    ms = 2000 - 4 * (last_idx - i)   # i=last_idx -> 2000; i=0 -> 2000-4*(n_frames-1)
    ts_label = f"{ms}ms"

    draw = ImageDraw.Draw(merged)
    try:
        bbox = draw.textbbox((0, 0), ts_label, font=font)
        text_w = bbox[2] - bbox[0]
        text_h = bbox[3] - bbox[1]
    except AttributeError:
        text_w, text_h = font.getsize(ts_label)

    pad = 20
    tx = merged.width - text_w - pad
    ty = pad
    draw_text_with_outline(draw, (tx, ty), ts_label, font=font, fill='white', outline='black', width=2)

    merged.save(os.path.join(outdir, f'merged_{i:03d}.png'))

print(f"All merged frames saved in {outdir}")
```
